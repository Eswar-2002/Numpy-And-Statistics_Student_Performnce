```python
df = pd.read_csv("Student_Performance.csv")
```

```python
data = {
    'student_id': range(1, 21),
    'age': np.random.randint(14, 20, 20),
    'gender': np.random.choice(['male', 'female', 'other'], 20),
    'school_type': np.random.choice(['public', 'private'], 20),
    'parent_education': np.random.choice(['high school', 'graduate', 'post graduate', 'phd'], 20),
    'study_hours': np.random.uniform(0.5, 8.0, 20),
    'attendance_percentage': np.random.uniform(50.0, 100.0, 20),
    'internet_access': np.random.choice(['yes', 'no'], 20),
    'math_score': np.random.uniform(20, 100, 20),
    'science_score': np.random.uniform(20, 100, 20),
    'english_score': np.random.uniform(20, 100, 20),
    'overall_score': np.random.uniform(20, 100, 20),
    'final_grade': np.random.choice(['a', 'b', 'c', 'd', 'e', 'f'], 20)
}
```

```python
df = pd.DataFrame(data)
```

```python
numeric_cols = ['math_score', 'science_score', 'english_score', 'overall_score', 'study_hours', 'attendance_percentage']
```

```python
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors='coerce')
```

```python
df.head()
```

```python
df.loc[0, 'overall_score'] = 5
df.loc[1, 'study_hours'] = 25
```

```python
numeric_cols = ['math_score', 'science_score', 'english_score', 'overall_score', 'study_hours', 'attendance_percentage']
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors='coerce')
```

Z-SCORE NORMALIZATION

```python
z_scores = df[numeric_cols].apply(stats.zscore)
df_normalized = pd.DataFrame(z_scores, columns=numeric_cols)
```

```python
print(df_normalized.head())
print(f"\nMean of normalized data (should be close to 0):\n{df_normalized.mean()}")
print(f"\nStd Dev of normalized data (should be 1):\n{df_normalized.std()}")
```

 OUTLIER DETECTION

```python
threshold = 3
outlier_mask = (np.abs(df_normalized) > threshold).any(axis=1)
outliers_df = df[outlier_mask]
```

```python
print(f"Number of Outliers Detected: {len(outliers_df)}")
if len(outliers_df) > 0:
  outlier_indices = outliers_df.index
```

```python
analysis_cols = ['student_id'] + numeric_cols
display_df = outliers_df[analysis_cols].copy()
```

```python
for col in numeric_cols:
        display_df[f'{col}_z'] = df_normalized.loc[outlier_indices, col]

print(display_df)
```

Basic Statistics

```python
stats = df[numeric_cols].describe().T
```

```python
stats['variance'] = df[numeric_cols].var()
```

```python
# Mean, Median, Std Dev, Variance, Min, Max
requested_metrics = stats[['mean', '50%', 'std', 'variance', 'min', 'max']]
requested_metrics.rename(columns={'50%': 'median'}, inplace=True)
```

```python
print(requested_metrics)
```

Correlation Matrix

```python
corr_matrix = df[numeric_cols].corr()
print(corr_matrix)
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix Heatmap')
plt.show()
```

PROBABILITY EXAMPLE

```python
prob_internet = len(df[df['internet_access'] == 'yes']) / len(df)
print(f"P(Internet Access): {prob_internet:.2%}")
```

```python
prob_high_grade = len(df[df['final_grade'].isin(['a', 'b'])]) / len(df)
print(f"P(Grade A or B): {prob_high_grade:.2%}")
```

```python
# A. Bar Plot: Average Overall Score by School Type
plt.figure(figsize=(8, 5))
sns.barplot(x='school_type', y='overall_score', data=df, errorbar=None, palette='viridis')
plt.title('Average Overall Score by School Type')
plt.ylabel('Average Overall Score')
plt.show()
```

```python
# B. Line Plot: Trend of Overall Score across Students (Sorted by ID)
plt.figure(figsize=(10, 5))
# Sorting to see the sequence of students
df_sorted = df.sort_values('student_id')
plt.plot(df_sorted['student_id'], df_sorted['overall_score'], marker='o', linestyle='-')
plt.title('Student Performance Trend (Overall Score)')
plt.xlabel('Student ID')
plt.ylabel('Overall Score')
plt.grid(True)
plt.show()
```

```python
# Scatter Plot: Study Hours (Original) vs Z-Score of Overall Score
plt.figure(figsize=(8, 5))
plt.scatter(df['study_hours'], df_normalized['overall_score'], c='blue', label='Normal Data')
plt.scatter(outliers_df['study_hours'], df_normalized.loc[outliers_df.index, 'overall_score'], c='red', s=100, label='Outliers')
plt.title('Outlier Detection: Study Hours vs Z-Score of Overall Score')
plt.xlabel('Study Hours')
plt.ylabel('Z-Score of Overall Score')
plt.axhline(y=3, color='r', linestyle='--')
plt.axhline(y=-3, color='r', linestyle='--')
plt.legend()
plt.show()
```

```python
# Boxplot of Z-Scores to visually identify outliers
plt.figure(figsize=(12, 6))
sns.boxplot(data=df_normalized, palette='Set2')
plt.axhline(y=3, color='r', linestyle='--', label='Threshold (+3)')
plt.axhline(y=-3, color='r', linestyle='--', label='Threshold (-3)')
plt.title('Boxplot of Z-Scores (Outlier Detection)')
plt.ylabel('Z-Score')
plt.legend()
plt.show()
```
